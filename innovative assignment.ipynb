```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score , classification_report
from sklearn.neighbors import KNeighborsClassifier


df = pd.read_csv("engineering colleges in India.csv")
df_ = df.drop([1735, 2871, 2870, 4278, 4495], axis=0)
df15 = df_.copy()
df15['Average Fees'] = df15['Average Fees'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)

mean_fees = df15.loc[~df15["Average Fees"].isna(), "Average Fees"].mean()
df15.loc[df15["Average Fees"].isna(), "Average Fees"]=mean_fees


df15["Fee Category"] = "0"
df15.loc[(df15["Average Fees"] >= 200000) & (df15["Average Fees"] <= 500000), "Fee Category"] = "1"
df15.loc[df15["Average Fees"] > 500000, "Fee Category"] = "2"


x = df15.iloc[:, 14:-1].values  
y = df15.iloc[:, -1].astype(int).values 


x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y)


print("Shapes:", x_train.shape, y_train.shape, x_test.shape, y_test.shape)


nclasses = np.unique(y_train).shape[0]
print(nclasses)


pred = np.zeros(shape=(x_test.shape[0],), dtype=int)


k = int(input("Enter value of k: "))

for i in range(x_test.shape[0]):
    dist = np.sqrt(np.sum((x_train - x_test[i])**2, axis=1)) 
    kminid = np.argpartition(dist, k)[:k]  

    invdist = 1 / (dist + 1e-10)  
    denom = np.sum(invdist[kminid])

    classvotes = np.zeros(nclasses)  

    for j in range(k):
        classvotes[int(y_train[kminid[j]])] += invdist[kminid[j]]

    classvotes /= denom  
    pred[i] = np.argmax(classvotes)  
    print(pred[i])


model=KNeighborsClassifier(n_neighbors=k,weights="distance")
model.fit(x_train,y_train)#training
pred=model.predict(x_test)#prediction
print(pred)
accuracy=accuracy_score(y_test,pred)
print("accuracy:",accuracy)
print(classification_report(y_test,pred))
```

```python
import pandas as pd
import numpy as np
import math
from sklearn.model_selection  import train_test_split

df = pd.read_csv("engineering colleges in India.csv")
df_ = df.drop([1735, 2871, 2870, 4278, 4495], axis=0)
df20 = df_.copy()

df20['Average Fees'] = df20['Average Fees'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
mean_fees = df20.loc[~df20["Average Fees"].isna(), "Average Fees"].mean()
df20.loc[df20["Average Fees"].isna(), "Average Fees"]=mean_fees

df20['Total Student Enrollments'] = df20['Total Student Enrollments'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
mean_enrol=df20.loc[~df20["Total Student Enrollments"].isna(),"Total Student Enrollments"].mean()
df20.loc[df20.loc[:,"Total Student Enrollments"].isna(),"Total Student Enrollments"]=mean_enrol

df20['Campus Size'] = df['Campus Size'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)
mean_size=df20.loc[~df20["Campus Size"].isna(),"Campus Size"].mean()
df20.loc[df20.loc[:,"Campus Size"].isna(),"Campus Size"]=mean_size

x=df20[:,2:5]
y=df20[:,-1]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2 ,stratify=y)
print("shapes=",x_train.shape,y_train.shape,x_test.shape,y_test.shape)
dist=np.zeros(shape=x_train.shape[0])
pred=np.zeros(shape=x_test.shape[0]) 
k=int(input("enter value of k"))
for i in range (x_test.shape[0]):
    dist = np.sqrt(np.sum((x_train - x_test[i]) ** 2, axis=1))
    kminid=np.argpartition(dist,k)[0:k] 
    invdist=1/(dist+10e-20)  
    denom=sum(invdist[kminid]) 
    pred[i]=np.dot(invdist[kminid]/denom,y_train[kminid])
   
print(pred)
def MAE(pred,y_test):
    return np.mean((pred-y_test))
def MSE(pred,y_test):
    return np.mean((pred-y_test)**2)
def MAPE(pred, y_test):
    epsilon = 1e-10 
    return np.mean(np.abs((y_test - pred) / (y_test + epsilon))) * 100
def RMSE(pred,y_test):
    return np.sqrt(np.mean((y_test-pred)**2))

mae=MAE(pred,y_test)
mse=MSE(pred,y_test) 
rmse=RMSE(pred,y_test)
mape=MAPE(pred,y_test)

print("MAE",mae)
print("MSE",mse)
print("MAPE",mape)
print("RMSE",rmse)
```
